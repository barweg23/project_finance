```python
class TableFullName:
    stocks_silver = "databricksformula1.default.stocks_silver"
    stocks_gold_month = "databricksformula1.default.stocks_gold_month"
    stocks_gold_week = "databricksformula1.default.stocks_gold_week"
    stocks_gold_year = "databricksformula1.default.stocks_gold_year"
```

```python
TableFullName.stocks_silver 
```

```python
sql_year_query = f"""
SELECT mean(close) close_mean, avg(high) high_mean, avg(low) low_mean, avg(open) open_mean, year(date) year, ticker 
FROM 
{TableFullName.stocks_silver}
group by year(date), ticker"""
```

```python
sql_month_query = f"""SELECT mean(close) close_mean, avg(high) high_mean, avg(low) low_mean, avg(open) open_mean, year(date) year, month(date) month, ticker 
FROM 
{TableFullName.stocks_silver}
group by year(date), month(date), ticker"""
```

```python
sql_weekly_query = f"""SELECT mean(close) close_mean, avg(high) high_mean, avg(low) low_mean, avg(open) open_mean, year(date) year, month(date) month, weekofyear(date) week, ticker 
FROM {TableFullName.stocks_silver}
group by year(date), month(date), weekofyear(date), ticker"""
```

```python
spark.sql(sql_year_query).write.mode("overwrite").saveAsTable(TableFullName.stocks_gold_year)
spark.sql(sql_month_query).write.mode("overwrite").saveAsTable(TableFullName.stocks_gold_month)
spark.sql(sql_weekly_query).write.mode("overwrite").saveAsTable(TableFullName.stocks_gold_week)
```

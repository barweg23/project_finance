```python
df = spark.table("databricksformula1.default.stocks")
```

```python
from pyspark.sql import functions as sf
from pyspark.sql import Window

df = df.where(sf.col("ticker").isNotNull())
w = Window.partitionBy("ticker").orderBy(sf.col("created_timestamp").desc())
df = df.withColumn("rank", sf.rank().over(w))
```

```python
df = df.where(sf.col("rank") == 1)
df.write.mode("overwrite").saveAsTable("databricksformula1.default.stocks_silver")
```
